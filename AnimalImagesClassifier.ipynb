```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow import keras
from keras import Sequential
from keras.layers import Flatten,Conv2D,MaxPooling2D,Dense
from keras import layers
```

```python
from pathlib import Path
project_dir = Path(r"D:\animal Dec")
train_ds=keras.utils.image_dataset_from_directory(
    directory=str(project_dir /"train"),
    labels="inferred",
    label_mode="int",
    class_names=None,
    batch_size=32,
    image_size=(128,128),
    color_mode='rgb',
)
valid_ds=keras.utils.image_dataset_from_directory(
    directory=str(project_dir/"test"),
    labels="inferred",
    label_mode="int",
    class_names=None,
    batch_size=32,
    image_size=(128,128),
    color_mode='rgb',
)
```

```python
def process(image,label):
    image=tf.cast(image/255,tf.float32)
    return image,label

train_ds=train_ds.map(process)
valid_ds=valid_ds.map(process)
```

```python
train_ds = train_ds.prefetch(buffer_size=tf.data.AUTOTUNE)
valid_ds = valid_ds.prefetch(buffer_size=tf.data.AUTOTUNE)
```

```python
data_augmentation = keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.1),
    layers.RandomZoom(0.1),
])
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense

model = Sequential()

model.add(data_augmentation)

model.add(Conv2D(32, (3,3), activation='relu', input_shape=(128,128,3)))
model.add(MaxPooling2D(2,2))

model.add(Conv2D(64, (3,3), activation='relu'))
model.add(MaxPooling2D(2,2))

model.add(Conv2D(128, (3,3), activation='relu'))
model.add(MaxPooling2D(2,2))

model.add(Conv2D(256,(3,3),activation='relu'))
model.add(MaxPooling2D(2,2))

model.add(Flatten())

model.add(Dense(128, activation='relu'))
model.add(Dense(64, activation='relu'))

# 4 classes: dog, cat, cow, background
model.add(Dense(4, activation='softmax'))

model.summary()
```

```python
model.summary()
```

```python
train_ds = train_ds.cache().prefetch(buffer_size=tf.data.AUTOTUNE)
```

```python
from tensorflow.keras.optimizers import Adam
optimizer=Adam(learning_rate=0.03)
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

```python
model.fit(
    train_ds,
    epochs=20,
    validation_data=valid_ds
)
```

```python
model.save("animal_images_classifier.keras")
```

```python
from tensorflow.keras.models import load_model

model = load_model("animal_images_classifier.keras")

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
import numpy as np
import cv2

# class names (IMPORTANT: same order as training folders)
class_names = ['dog', 'cat', 'elephant', 'sheep']

def predict_image(model, image_path):
    # 1. Image load karo
    img = cv2.imread(image_path)
    
    # 2. Resize (same as training)
    img = cv2.resize(img, (128, 128))
    
    # 3. Normalize
    img = img / 255.0
    
    # 4. Batch dimension add karo
    img = np.expand_dims(img, axis=0)
    
    # 5. Prediction
    preds = model.predict(img, verbose=0)
    
    # 6. Result nikaalo
    class_id = np.argmax(preds)
    confidence = np.max(preds)
    
    label = class_names[class_id]
    
    return label, confidence
```

```python
label, conf = predict_image(model, "Golden+Retrievers+dans+pet+care.jpeg")

print("Prediction:", label)
print("Confidence:", conf)
```
