```python
import albumentations as A
from albumentations.pytorch import ToTensorV2
def get_train_transform():
    return A.Compose([
        A.Resize(600,600),
        A.HorizontalFlip(p=0.3),
        A.VerticalFlip(p=0.3),
        A.RandomBrightnessContrast(p=0.1),
        A.ColorJitter(p=0.1)
    ],bbox_params=A.BboxParams(format='coco', label_fields=['labels']))

def get_test_transform():
    return A.Compose([
        A.Resize(600,600)
    ],bbox_params=A.BboxParams(format='coco', label_fields=['labels']))
```

```python
import os
import cv2
import torch
from torch.utils.data import Dataset
from pycocotools.coco import COCO


class AquariumDetection(Dataset):
    def __init__(self, root, split="train", transform=None):
        """
        Args:
            root: Dataset root directory.
            split: 'train', 'valid', or 'test'
            transform: Albumentations transform
        """
        self.root = root
        self.split = split
        self.transform = transform

        annotation_file = os.path.join(
            root,
            split,
            "_annotations.coco.json"
        )

        self.coco = COCO(annotation_file)

        # Keep only images that contain annotations
        self.ids = [
            img_id
            for img_id in sorted(self.coco.imgs.keys())
            if len(self.coco.getAnnIds(imgIds=img_id)) > 0
        ]

    def load_image(self, img_id):
        img_info = self.coco.loadImgs(img_id)[0]
        img_path = os.path.join(
            self.root,
            self.split,
            img_info["file_name"]
        )

        image = cv2.imread(img_path)
        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

        return image

    def __getitem__(self, index):

        img_id = self.ids[index]

        image = self.load_image(img_id)

        ann_ids = self.coco.getAnnIds(imgIds=img_id)
        annotations = self.coco.loadAnns(ann_ids)

        # COCO boxes: [x, y, width, height]
        boxes = [ann["bbox"] for ann in annotations]
        labels = [ann["category_id"] for ann in annotations]
        iscrowd = [ann.get("iscrowd", 0) for ann in annotations]

        if self.transform:
            transformed = self.transform(
                image=image,
                bboxes=boxes,
                labels=labels
            )

            image = transformed["image"]
            boxes = transformed["bboxes"]
            labels = transformed["labels"]

        # Manually convert image to float32 tensor and permute dimensions after Albumentations
        image = torch.from_numpy(image).permute(2, 0, 1).float() / 255.0

        # Convert boxes to Pascal VOC format
        boxes_tensor = torch.tensor(
            [[x, y, x + w, y + h] for x, y, w, h in boxes],
            dtype=torch.float32
        )

        labels_tensor = torch.tensor(labels, dtype=torch.int64)
        iscrowd_tensor = torch.tensor(iscrowd, dtype=torch.int64)

        area = (
            (boxes_tensor[:, 3] - boxes_tensor[:, 1]) *
            (boxes_tensor[:, 2] - boxes_tensor[:, 0])
        )

        target = {
            "boxes": boxes_tensor,
            "labels": labels_tensor,
            "image_id": torch.tensor([img_id]),
            "area": area,
            "iscrowd": iscrowd_tensor
        }

        return image, target

    def __len__(self):
        return len(self.ids)
```

```python
dataset_dir="/content/aquarium-dataset/Aquarium Combined"
train_dataset=AquariumDetection(root=dataset_dir,transform=get_train_transform())
```

```python
import torchvision
from torchvision import models
from torchvision.models.detection.faster_rcnn import FastRCNNPredictor
model = models.detection.fasterrcnn_mobilenet_v3_large_fpn(pretrained=True)
model.roi_heads.box_predictor = models.detection.faster_rcnn.FastRCNNPredictor(
    model.roi_heads.box_predictor.cls_score.in_features,
    len(train_dataset.coco.cats)
)
```

```python
import torch.optim as optim
params = [p for p in model.parameters() if p.requires_grad]

optimizer = optim.SGD(
    params,
    lr=0.005,
    momentum=0.9,
    weight_decay=0.0005
)
```

```python
from torch.utils.data import DataLoader
from torchmetrics.detection import MeanAveragePrecision

DEVICE = torch.device('cuda') if torch.cuda.is_available() else torch.device('cpu')

valid_dataset = AquariumDetection(root=dataset_dir, split="valid", transform=get_test_transform())

def collate_fn(batch):
    return tuple(zip(*batch))

train_loader = DataLoader(
    train_dataset,
    batch_size=4,
    shuffle=True,
    num_workers=2,
    collate_fn=collate_fn
)

valid_loader = DataLoader(
    valid_dataset,
    batch_size=4,
    shuffle=False,
    num_workers=2,
    collate_fn=collate_fn
)

# Initialize lists to store metrics for plotting
train_losses_classifier = []
train_losses_box_reg = []
train_losses_objectness = []
train_losses_rpn_box_reg = []
train_avg_losses = []

valid_map_scores = []
valid_map_50_scores = []
valid_map_75_scores = []
valid_mar_100_scores = []

# Move model to device
model.to(DEVICE)
```

```python
import os
import cv2
import torch
from torch.utils.data import Dataset
from pycocotools.coco import COCO
import torchvision
from torchvision import models
from torchvision.models.detection.faster_rcnn import FastRCNNPredictor

# Import tqdm for progress bars
from tqdm.notebook import tqdm

def train_one_epoch(model, curr_epoch_no):
    """
    Trains the model for one epoch and evaluates it on the validation set.

    Args:
        model: The Faster R-CNN model.
        curr_epoch_no: Current epoch number.
    """

    # ----------------------
    # TRAINING PHASE
    # ----------------------

    model.train()  # Set model to training mode
    tot_epoch_loss = 0  # Track total loss for averaging later

    # Initialize dictionary to accumulate individual losses
    loss_dict_accum = {
        'loss_classifier': 0,
        'loss_box_reg': 0,
        'loss_objectness': 0,
        'loss_rpn_box_reg': 0
    }

    num_of_batches = len(train_loader)

    print(f"\n\nRunning epoch {curr_epoch_no + 1}...\n")

    for batch_images, batch_targets in tqdm(train_loader):
        # Move images and targets to device (GPU/CPU)
        batch_images = [img.to(DEVICE) for img in batch_images]
        batch_targets = [{key: value.to(DEVICE) for key, value in target.items()} for target in batch_targets]

        # Compute loss
        loss_dict = model(batch_images, batch_targets)
        total_loss = sum(loss for loss in loss_dict.values())

        # Accumulate total epoch loss
        tot_epoch_loss += total_loss.item()

        # Accumulate losses for averaging
        for key, val in loss_dict.items():
            loss_dict_accum[key] += val.item()

        # Backpropagation & Optimization
        optimizer.zero_grad()
        total_loss.backward()
        optimizer.step()

    # Compute average loss for the epoch
    avg_epoch_loss = tot_epoch_loss / num_of_batches
    avg_loss_dict_accum = {key: val / num_of_batches for key, val in loss_dict_accum.items()}

    # Print training loss metrics
    print(f"(Training) Avg loss: {avg_epoch_loss}")
    for key, value in avg_loss_dict_accum.items():
        print(f"(Training) {key}: {value}")

    # Append losses for plotting later
    train_losses_classifier.append(avg_loss_dict_accum["loss_classifier"])
    train_losses_box_reg.append(avg_loss_dict_accum["loss_box_reg"])
    train_losses_objectness.append(avg_loss_dict_accum["loss_objectness"])
    train_losses_rpn_box_reg.append(avg_loss_dict_accum["loss_rpn_box_reg"])
    train_avg_losses.append(avg_epoch_loss)


    # ----------------------
    # VALIDATION PHASE
    # ----------------------

    model.eval()  # Set model to evaluation mode
    metric = MeanAveragePrecision(iou_type="bbox")  # Initialize metric

    with torch.no_grad():  # Disable gradient computation for validation
        for batch_images, batch_targets in tqdm(valid_loader):
            # Move images and targets to device
            batch_images = [img.to(DEVICE) for img in batch_images]
            batch_targets = [{key: value.to(DEVICE) for key, value in target.items()} for target in batch_targets]

            # Forward pass (get predictions)
            outputs = model(batch_images)

            # Convert model outputs to expected metric format
            predictions = [
                {
                    "boxes": output['boxes'].detach().cpu(),
                    "scores": output['scores'].detach().cpu(),
                    "labels": output['labels'].detach().cpu()
                }
                for output in outputs
            ]

            targets = [
                {
                    "boxes": target['boxes'].detach().cpu(),
                    "labels": target['labels'].detach().cpu()
                }
                for target in batch_targets
            ]

            # Update evaluation metric
            metric.update(predictions, targets)

    # Compute validation metrics
    metric_result = metric.compute()

    # Print validation metrics
    print(f"\n(Validation) mAP @ IoU 0.5:0.95: {metric_result['map']}")
    print(f"(Validation) mAP @ IoU 0.5: {metric_result['map_50']}")
    print(f"(Validation) mAP @ IoU 0.75: {metric_result['map_75']}")
    print(f"(Validation) mAR @ 100 detections per image: {metric_result['mar_100']}")

    # Append validation metrics for tracking progress
    valid_map_scores.append(metric_result['map'].item())
    valid_map_50_scores.append(metric_result['map_50'].item())
    valid_map_75_scores.append(metric_result['map_75'].item())
    valid_mar_100_scores.append(metric_result['mar_100'].item())
```

```python
for epoch in range(0, 10):
    train_one_epoch(model, epoch)
```

```python
def plot_images_with_predictions(original_image, predictions, labels_map):
    plt.figure(figsize=(12, 6))

    # Display original image
    plt.subplot(1, 2, 1)
    plt.imshow(original_image)
    plt.title("Original Image")
    plt.axis("off")

    # Display image with predictions
    plt.subplot(1, 2, 2)
    plt.imshow(original_image)
    plt.title("Predicted Objects")

    # Draw bounding boxes and labels on the predicted image
    for box, label, score in zip(
        predictions["boxes"],
        predictions["labels"],
        predictions["scores"]
    ):
        if score >= 0.5:  # Confidence threshold
            box = box.to(torch.int).tolist()

            plt.gca().add_patch(
                plt.Rectangle(
                    (box[0], box[1]),
                    box[2] - box[0],
                    box[3] - box[1],
                    linewidth=2,
                    edgecolor="red",
                    facecolor="none"
                )
            )

            plt.text(
                box[0],
                box[1] - 5,
                f"{labels_map[label.item()]}: {score:.2f}",
                color="red",
                fontsize=12
            )

    plt.axis("off")
    plt.show()
```

```python
test_dataset = AquariumDetection(
    root=dataset_dir,
    split="test",
    transform=get_test_transform()
)
from torch.utils.data import DataLoader

def collate_fn(batch):
    return tuple(zip(*batch))

test_loader = DataLoader(
    test_dataset,
    batch_size=4,
    shuffle=False,
    collate_fn=collate_fn
)
```

```python
import torchvision.transforms.functional as F
def evaluate(model, test_loader, device, labels_map):
    model.eval()

    with torch.no_grad():
        for images, targets in test_loader:

            images = [img.to(device) for img in images]
            outputs = model(images)

            # Process each image in the batch
            for i, output in enumerate(outputs):

                original_image = F.to_pil_image(images[i].cpu())

                # Move tensors to CPU for visualization
                predictions = {
                    "boxes": output["boxes"].cpu(),
                    "labels": output["labels"].cpu(),
                    "scores": output["scores"].cpu()
                }

                # Display the original and predicted image side by side
                plot_images_with_predictions(
                    original_image,
                    predictions,
                    labels_map
                )

            # Break after first batch to limit visualization to a few images
            break
```

```python
labels_map = {
    1: "fish",
    2: "jellyfish",
    3: "penguin",
    4: "puffin",
    5: "shark",
    6: "starfish",
    7: "stingray",
}
```

```python
evaluate(
    model,
    test_loader,
    DEVICE,
    labels_map
)
```
